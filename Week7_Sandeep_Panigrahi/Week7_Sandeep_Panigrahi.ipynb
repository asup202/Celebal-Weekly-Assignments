```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("Ch13_ML Implementation.pdf")

documents = loader.load()
```

```python
print("Total Pages:", len(documents))
```

```python
print(type(documents))
```

```python
print(type(documents[0]))
```

```python
print(documents[0].metadata)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = text_splitter.split_documents(documents)
```

```python
print("Total Chunks:", len(chunks))
print(chunks[0].page_content)
print(chunks[1].page_content)
print(chunks[0].metadata)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
embedding = embedding_model.embed_query(chunks[0].page_content)
```

```python
print(type(embedding))
print(len(embedding))
print(embedding[:10])
```

```python
from langchain_community.vectorstores import FAISS

vector_db = FAISS.from_documents(
    documents=chunks,
    embedding=embedding_model
)
```

```python
print(type(vector_db))
```

```python
vector_db.save_local("faiss_index")
```

```python
query = "Which machine learning algorithm is used?"
```

```python
results = vector_db.similarity_search(query, k=3)
```

```python
for i, doc in enumerate(results, 1):
    print("=" * 80)
    print(f"Result {i}")
    print(f"Page: {doc.metadata['page'] + 1}")
    print(doc.page_content)
    print()
```
